### Example 1 – PCA

**PCA** looks for the directions (components) with the most variance in the data.

The projection preserves as much information as possible, but in a lower-dimensional space.


```python
# Principal Component Analysis (PCA)

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 1. Load the data
data = load_iris()

X = data.data
y = data.target

# 2. Scale the data (important for PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Run PCA and reduce to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 4. Visualize the PCA projection
plt.figure(figsize=(6, 4))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap="viridis")

plt.title("PCA – 2D projection")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```

### Example 2 – t-SNE

**t-SNE** tries to preserve local structures: points that were close to each other in the original space also remain close to each other in 2D.

It is particularly useful for discovering and visualizing clusters.


```python
# t-Distributed Stochastic Neighbor Embedding (t-SNE)

from sklearn.manifold import TSNE

# 1. Run t-SNE
tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne = tsne.fit_transform(X_scaled)

# 2. Visualize the t-SNE projection
plt.figure(figsize=(6, 4))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap="plasma")

plt.title("t-SNE – 2D projection")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```

### Example 3 – UMAP

**UMAP** can handle larger datasets efficiently and tries to preserve local structure, while often also making more of the global structure visible than t-SNE.

It is therefore useful for exploratory visualizations.

> **Note:** for this example, you need the `umap-learn` package.


```python
# Uniform Manifold Approximation and Projection (UMAP)

import umap

# 1. Run UMAP
umap_model = umap.UMAP(
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap = umap_model.fit_transform(X_scaled)

# 2. Visualize the UMAP projection
plt.figure(figsize=(6, 4))
plt.scatter(X_umap[:, 0], X_umap[:, 1], c=y, cmap="cool")

plt.title("UMAP – 2D projection")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```
