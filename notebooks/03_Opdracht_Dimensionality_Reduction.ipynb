### Voorbeeld 1 – PCA

**PCA** zoekt naar de richtingen (componenten) met de meeste variantie in de data.

De projectie behoudt zoveel mogelijk informatie, maar in een lager-dimensionale ruimte.


```python
# Principal Component Analysis (PCA)

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 1. Data laden
data = load_iris()

X = data.data
y = data.target

# 2. Schalen van de data (belangrijk voor PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. PCA uitvoeren en terugbrengen naar 2 dimensies
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 4. Visualiseren van de PCA-projectie
plt.figure(figsize=(6, 4))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap="viridis")

plt.title("PCA – 2D projectie")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```

### Voorbeeld 2 – t-SNE

**t-SNE** probeert lokale structuren te behouden: punten die dicht bij elkaar lagen in de originele ruimte, blijven ook dicht bij elkaar in 2D.

Het is vooral geschikt voor het ontdekken en visualiseren van clusters.


```python
# t-Distributed Stochastic Neighbor Embedding (t-SNE)

from sklearn.manifold import TSNE

# 1. t-SNE uitvoeren
tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne = tsne.fit_transform(X_scaled)

# 2. Visualiseren van de t-SNE-projectie
plt.figure(figsize=(6, 4))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap="plasma")

plt.title("t-SNE – 2D projectie")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```

### Voorbeeld 3 – UMAP

**UMAP** kan efficiënt omgaan met grotere datasets en probeert lokale structuur te behouden, terwijl het vaak ook meer van de globale structuur zichtbaar houdt dan t-SNE.

Het is daarom handig voor verkennende visualisaties.

> **Let op:** voor dit voorbeeld heb je het pakket `umap-learn` nodig.


```python
# Uniform Manifold Approximation and Projection (UMAP)

import umap

# 1. UMAP uitvoeren
umap_model = umap.UMAP(
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap = umap_model.fit_transform(X_scaled)

# 2. Visualiseren van de UMAP-projectie
plt.figure(figsize=(6, 4))
plt.scatter(X_umap[:, 0], X_umap[:, 1], c=y, cmap="cool")

plt.title("UMAP – 2D projectie")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```
